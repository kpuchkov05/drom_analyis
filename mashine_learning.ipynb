```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
import statsmodels.api as sm
```

Подготовим данные для обучения модели

```python
df = pd.read_csv('data\priora_data.csv')
pd.set_option('display.max_columns', None)
df
```

```python
df = df[df['price_score'].isna() == False]
df = df[df['price_score'] != 'без оценки']
```

```python
df['price_score'] = df['price_score'].apply(lambda x: 0.25 if x == 'отличная цена' else 0.5 if x == 'хорошая цена' else 0.75 if x == 'нормальная цена' else 1)
```

```python
df['price'] = df['price'].apply(lambda x: int(''.join(x.split()[:-1])))
```

```python
df.drop(columns='Привод', inplace= True)
```

```python
df.rename(columns = {
        'title': 'year',
        'Двигатель': 'engine',
        'Мощность': 'power',
        'Коробка передач': 'box',
        'Тип кузова': 'bodywork',
        'Цвет': 'color',
        'Пробег': 'mileage',
        'Владелец': 'owner',
        'Руль': 'wheel',
        'Поколение': 'generation',
        'Комплектация': 'equipment'
                    }, inplace = True)
```

```python
df.drop(columns='Владельцы', inplace= True)
```

```python
df['year'] = df['year'].apply(lambda x: int(x.split()[3]))
df['power'] = df['power'].apply(lambda x: int(x.split()[1]))
df['engine'] = df['engine'].apply(lambda x: float(x.split()[2]))
```

```python
df['box'] = df['box'].apply(lambda x: str(x).split()[-1])
df['bodywork'] = df['bodywork'].apply(lambda x: ' '.join(x.split()[2:]))
```

```python
df['color'] = df['color'].apply(lambda x: ' '.join(str(x).split()[1:]))
df['mileage']= df['mileage'].apply(lambda x: float(''.join(x.split()[1:-1])))
df.drop(columns= 'wheel', inplace=True)
```

```python
df['generation'] = df['generation'].apply(lambda x: 2 if 'рестайлинг' in str(x) else 1)
df['pts'] = df['pts'].apply(lambda x: 0 if 'не' in str(x) else 1)
df['registration'] = df['registration'].apply(
    lambda x: int(str(x).split()[0]) if not pd.isna(x) else None
)
df['gibdd'] = df['gibdd'].apply(lambda x: 0 if 'Не' in str(x) else 1)
df['ur_face'] = df['ur_face'].apply(lambda x: 1 if 'Не' in str(x) else 0)
```

```python
df.drop(columns= 'limitations', inplace=True)
df.drop(columns= 'search', inplace=True)
df.drop(columns='box', inplace=True)
```

```python
df['period'] = df['period'].apply(lambda x: ' '.join(str(x).split()[-2:]))
```

```python
def parse_period(x):
    if pd.isna(x):
        return None
    try:
        parts = str(x).split()
        value = float(parts[0])
        unit = parts[1]
        if 'месяц' in unit:
            return 0.5
        else:
            return int(value)
    except:
        return None

df['period'] = df['period'].apply(parse_period)
```

```python
df['period'].value_counts()
```

```python
df['equipment'] = df['equipment'].apply(lambda x: ' '.join(str(x).split()[3:4]))
```

```python
df['equipment'] = df['equipment'].apply(lambda x: x.replace('', 'Норма') if len(x) == 0 else x)
df
```

```python
df = df[df['registration'].isna() == False]
```

```python
df.info()
```

```python
df.drop(columns= 'engine', inplace= True)
```

Данные почти готовы для обучения модели. Для кодирования категориальных признаков используем Onehot Encoding

```python
y = df['price_score']
df.drop(columns= 'price_score', inplace=True)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

categorical_cols = ['bodywork', 'color', 'equipment']
X = df

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ],
    remainder='passthrough'
)

X_transformed = preprocessor.fit_transform(X)

X = pd.DataFrame(X_transformed, columns=preprocessor.get_feature_names_out())
y = y.reset_index(drop=True)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = sm.OLS(y_train, sm.add_constant(X_train))
results = model.fit()
display(results.summary())
```

Попробуем случайный лес, так как признаков много, и похоже, что есть нелинейная зависимость

```python
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(n_estimators=150,
                                max_depth= 20,
                                max_features= None, 
                                min_samples_leaf= 2, 
                                min_samples_split= 2, 
                                random_state=42)
model.fit(X, y)
print(model.score(X, y))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.37, random_state=42)

model.fit(X_train, y_train)

test_score = model.score(X_test, y_test)
print("r^2 на тестовой выборке:", test_score)
```

Попробуем отбор признаков

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_estimators': [100, 200, 300, 500],
    'max_depth': [10, 20, 30, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 'log2', None]
}


rf = RandomForestRegressor(random_state=42)

grid_search = GridSearchCV(estimator=rf, param_grid=param_grid, cv=5, n_jobs=-1)

grid_search.fit(X_train, y_train)

print("Лучшие параметры:", grid_search.best_params_)

best_rf = grid_search.best_estimator_
print(f'R^2 на тестовых данных: {best_rf.score(X_test, y_test)}')
```
