```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.support.ui import WebDriverWait
import time
import random
import pandas as pd
import numpy as np
```

```python
pd.set_option('display.max_columns', None)
```

```python
ls = []
browser = webdriver.Chrome()
for i in range(16):
    browser.get('https://auto.drom.ru/lada/priora/page' + str(i) + '/?cid[]=92&rid[]=50')
    links = browser.find_elements(By.CLASS_NAME, '_1ioeqy90')
    for i in links:
        link = i.get_attribute('href')
        ls.append(link)
        print(link)
    time.sleep(random.uniform(1, 7))
    
browser.quit()
```

```python
data = []
browser = webdriver.Chrome()
for i in ls:
    params = {
                'title' : np.nan,
                'price': np.nan,
                'price_score': np.nan,
                'Двигатель': np.nan,
                'Мощность': np.nan,
                'Коробка передач': np.nan,
                'Привод': np.nan,
                'Тип кузова': np.nan,
                'Цвет': np.nan,
                'Пробег': np.nan,
                'Владельцы': np.nan,
                'Руль': np.nan,
                'Поколение': np.nan,
                'Комплектация': np.nan,
                'pts': np.nan, 
                'registration': np.nan, 
                'gibdd': np.nan, 
                'ur_face': np.nan, 
                'period': np.nan, 
                'search': np.nan, 
                'limitations': np.nan
            }
    browser.get(i)
    try:
        title = browser.find_element(By.CLASS_NAME, 'css-6tq1oz').text
        params['title'] = title
    except:
        price = None
    try:
        price = browser.find_element(By.CLASS_NAME, 'wb9m8q0').text
        params['price'] = price
    except:
        price = None
    try:
        price_score = browser.find_element(By.CSS_SELECTOR, "div[data-ga-stats-name='good_deal_mark']").text
        params['price_score'] = price_score
    except:
        price_score = None
        
    try:
        table = browser.find_elements(By.CLASS_NAME, 'css-10ib5jr')[1].text.split('\n')
        for i in table:
            for j in params.keys():
                if j in i:
                    params[j] = i
    except:
        table = None
    
    browser.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    try:
        table = browser.find_elements(By.CSS_SELECTOR, ".css-13qo6o5.eawu4md1")
    except:
        table = None
        
    try:
        pts = table[0].text
        params['pts'] = pts
    except:
        pts = None
        
    try:
        registration = table[1].text
        params['registration'] = registration
    except:
        registration = None
        
    try:
        table = browser.find_elements(By.CLASS_NAME, 'e1mhp2ux0')
    except:
        table = None
        
    try:
        gibdd = table[0].text
        params['gibdd'] = gibdd
    except:
        gibdd = None
        
    try:
        ur_face = table[1].text
        params['ur_face'] = ur_face
    except:
        ur_face = None
        
    try:
        period = table[2].text
        params['period'] = period
    except:
        period = None
        
    try:
        search = table[3].text
        params['search'] = search
    except:
        search = None
        
    try:
        limitations = table[4].text
        params['limitations'] = limitations
    except:
        limitations = None
    
    data.append(params)
    time.sleep(random.uniform(1, 5))
browser.quit()

df = pd.DataFrame(data)
df
df.to_csv('data/priora_data.csv', index=False)
```
